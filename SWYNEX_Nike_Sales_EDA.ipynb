```python
df.groupby("Product_Line")["Profit"].sum()
```

```python
df.groupby("Product_Line")["Profit"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
```

```python
df.groupby("Product_Line")["Profit"].sum().plot(kind = "bar",
                                                title = "Total Profit by Product Line")
plt.xlabel ("Product Line")
plt.ylabel("Total Profit")
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```
